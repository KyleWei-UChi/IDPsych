# IDPsych example

Set `DATA_DIR` to the folder holding one sub-folder per subject, and `OUT_DIR` to where figures and the JSON summary should go.

```python
from pathlib import Path
from IDPsych import dataIO, vis

DATA_DIR = Path('../data')        # <- edit
OUT_DIR = Path('../outputs')
OUT_DIR.mkdir(exist_ok=True)
sjID = ['403', '404', '405', '406', '408']
```

## Load all subjects

```python
allData = dataIO.allSubjectDict(DATA_DIR, sjID, elim=10, method=0)
{sj: len(d['fileName']) for sj, d in allData.items()}
```

```python
dataIO.saveDict(allData, OUT_DIR / 'primaryData')
```

## Figures

```python
vis.IDscat(allData, saveTo=OUT_DIR / 'thresholds.pdf');
```

```python
vis.hitRateHist(allData, saveTo=OUT_DIR / 'hitRate.pdf');
```

```python
vis.rBias(allData, saveTo=OUT_DIR / 'rBias.pdf');
```

## Compare threshold methods on one subject

```python
import numpy as np
for m in (0, 1, 2):
    d = dataIO.loadSubject(DATA_DIR / sjID[0], method=m)
    print('method', m, np.round(d['thresholdPC'], 1))
```
